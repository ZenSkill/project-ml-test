```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_excel(r'Precio_coches.xlsx')
```

```python
df.head()
df.info()
```

```python
df_X = df.drop('Precio', axis=1)
df_y = df['Precio']
```

```python
df_X = pd.get_dummies(df_X)
```

```python
model = linear_model.LinearRegression()
model.fit(df_X, df_y)
```

```python
predicciones = model.predict(df_X)
```

```python
plt.scatter(df_y, predicciones)
```

```python
plt.scatter(df['Potencia'], df['Precio'])
```

```python
X = df[['Potencia']]
y = df['Precio']
modelo_potencia = linear_model.LinearRegression()
modelo_potencia.fit(X, y)
```

```python
plt.scatter(df['Potencia'], df['Precio'])
plt.plot(df['Potencia'], modelo_potencia.predict(X))
```
