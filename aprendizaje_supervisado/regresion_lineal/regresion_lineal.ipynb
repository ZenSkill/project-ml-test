```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.DataFrame()
df["Area"] = [2600,3000,3600,4000]
df["Precio"] = [550000,565000,610000,736800]
```

```python
plt.scatter(df["Area"],df["Precio"])
```

```python
X = df[["Area"]]
```

```python
type(X)
```

```python
y = df["Precio"]
```

```python
model = linear_model.LinearRegression()
model.fit(X,y)
```

```python
model.predict([[3300]])
```

```python
plt.plot(df["Area"],model.predict(X))
```

```python
plt.scatter(df["Area"],df["Precio"])
plt.plot(df["Area"],model.predict(X))
```
